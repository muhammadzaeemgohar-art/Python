## 📝 Challenge 01: Multiplication Table Generator ⭐

**Task:** Create an automated multiplication utility that takes an integer input from the user and prints its multiplication table from factor `1` up to `10`.

---

### 📋 Requirements

1. **User Input:** Prompt the user to enter a target number (Integer).
2. **Loop Construction:** Utilize a `for` loop combined with the `range()` function to iterate cleanly from `1` through `10`.
3. **Calculation:** Multiply the user's input by the current loop iteration index on each pass.
4. **Output Formatting:** Print each multiplication step in a clean, aligned tabular format using f-strings (`f"{number} x {i} = {result}"`).

---

### 📥 Example Walkthrough

#### Target Input: `7`
* **Expected Output:**
```markdown

## 📝 Challenge 02: Sum of Even and Odd Numbers ⭐

**Task:** Write a program that takes a positive integer $N$ from the user and calculates the separate cumulative sums of all even and odd numbers from $1$ to $N$ using a loop.

---

### 📋 Requirements

1. **User Input:** Prompt the user to enter a positive integer $N$.
2. **Data Validation:** Ensure the entered number is greater than `0`. If not, display an appropriate message.
3. **Loop & Accumulation:** 
   * Iterate through all numbers starting from `1` up to $N$ (inclusive).
   * Use the modulo operator (`%`) to test if each number is even (`num % 2 == 0`) or odd.
   * Maintain two running accumulator variables: one for the sum of even numbers and one for the sum of odd numbers.
4. **Output Formatting:** Print the calculated total sum for both even and odd categories clearly.

---

### 📥 Example Walkthrough

#### Target Input: `10`
* **Calculation:**
  * **Evens:** $2 + 4 + 6 + 8 + 10 = 30$
  * **Odds:** $1 + 3 + 5 + 7 + 9 = 25$
* **Expected Output:**
```markdown
Even Sum = 30
Odd Sum = 25

````python
num=int(input("Enter a Number:"))
even=0
odd=0
if num>0:
    for n in range(1,num+1):
        if n%2==0:
            even+=n
        else:
            odd+=n
    print(f"Target Input:{num}")
    print(f"Even Sum={even}")
    print(f"Odd Sum={odd}")
else:
    print("Please Enter A Positive Number.")
````

## 📝 Challenge 03: Number Guessing Game ⭐⭐

**Task:** Build an interactive guessing game simulator using a loop where the user repeatedly attempts to guess a predefined secret number, receiving dynamic hints ("Too High!" or "Too Low!") until they find the correct value.

---

### 📋 Requirements

1. **Game Setup:** Store a fixed secret number in a variable (e.g., `secret_number = 27`) and initialize an attempt counter variable to `0`.
2. **Interactive Loop:** Use a `while` loop that continues running indefinitely until the user successfully guesses the secret number.
3. **User Input & Comparison:**
   * On each iteration, prompt the user to input an integer guess.
   * Increment the attempt counter by `1`.
   * Compare the user's guess against the secret number:
     * If the guess is **greater** than the secret number, print `"Too High!"`.
     * If the guess is **less** than the secret number, print `"Too Low!"`.
     * If the guess matches the secret number, exit the loop.
4. **Victory Output:** Print a congratulations message displaying the total number of attempts used to solve the challenge.

---

### 📥 Example Walkthrough

#### Game Session Run
* **Inputs:** `50` $\rightarrow$ `12` $\rightarrow$ `27`
* **Expected Output:**
```markdown
Guess the secret number: 50
Too High!

Guess the secret number: 12
Too Low!

Guess the secret number: 27
Congratulations!
Attempts: 3

````python
secret_number=27
attempts=0
while True:
    guess=int(input("Guess the Secret Number:"))
    if guess>secret_number:
        print("Too High!")
        attempts+=1
    elif guess<secret_number:
        print("Too Low!")
        attempts+=1
    elif guess==secret_number:
        attempts+=1
        print(f"Congratulations!")
        print(f"Attempts:{attempts}")
        break
````
