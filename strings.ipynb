## 📝 Challenge 01: String Information ⭐

**Task:** Write a Python program that accepts a sentence from the user and displays its foundational properties: the exact string entered, its total character count, and the specific characters at the very beginning and end.

---

### 📋 Requirements

1. **User Input:** Prompt the user to enter a string/sentence.
2. **String Operations:**
   * Use the built-in `len()` function to calculate the total number of characters (including spaces).
   * Use index `[0]` to extract the very first character of the string.
   * Use negative index `[-1]` to extract the very last character of the string.
3. **Display Format:** Print the original string, its length, first character, and last character on separate lines with clean alignment.

---

### 📥 Example Walkthrough

#### User Input:
```text
Enter a sentence: Python is awesome

Original String : Python is awesome
Length          : 17
First Character : P
Last Character  : e

````python
sentence=str(input("Enter a Sentence:"))
print(f"Original Sentence: {sentence}")
print(f"Length : {len(sentence)}")
print(f"First Character : {sentence[0]}")
print(f"Last Character : {sentence[-1]}")
````

## 📝 Challenge 02: String Slicing ⭐

**Task:** Write a Python program that prompts the user to enter a word, then uses string slicing to extract and display its first 3 characters, last 3 characters, and the reversed word.

---

### 📋 Requirements

1. **User Input:** Ask the user to enter a word.
2. **String Slicing Operations:**
   * Extract the first 3 characters using `word[:3]`.
   * Extract the last 3 characters using `word[-3:]`.
   * Reverse the entire word using step slicing `word[::-1]`.
3. **Display Format:** Print all three results on separate lines with clean alignment.

---

### 📥 Example Walkthrough

#### User Input:
```text
Enter a word: Programming

First 3 Characters : Pro
Last 3 Characters  : ing
Reverse            : gnimmargorP

````python
word=str(input("Enter a Word:")).title()
print(f"Original Word : {word}")
print(f"First  3 Character : {word[:3]}")
print(f"Last 3 Character : {word[-3:]}")
print(f"Reverse : {word[::-1]}")
````

## 📝 Challenge Project: Password & Text Analyzer ⭐⭐⭐⭐⭐
