## 📝 Challenge 06: Password Strength Checker ⭐⭐⭐

**Task:** Create a function named `check_password(password)` that evaluates a given password string against multiple security criteria and returns whether it is `"Strong Password"` or `"Weak Password"`.

---

### 📋 Requirements

1. **Function Definition:** Declare a function named `check_password(password)`.
2. **Parameters & Return:** The function must accept **one parameter** (`password`) and **return** a string: either `"Strong Password"` or `"Weak Password"`.
3. **Validation Rules (Must meet ALL 4 to be Strong):**
   * **Length:** Minimum of **8 characters** (`len(password) >= 8`).
   * **Digits:** Contains at least **one numeric digit** (`0-9`).
   * **Uppercase:** Contains at least **one uppercase letter** (`A-Z`).
   * **Lowercase:** Contains at least **one lowercase letter** (`a-z`).
4. **Execution:** Prompt the user for a password input, pass it to `check_password()`, and print the returned evaluation.

---

### 📥 Example Walkthroughs

#### Case A: Valid Password
* **Input:** `Python123`
* **Checks:** Length (9 $\ge$ 8) ✅ | Digit ('1') ✅ | Uppercase ('P') ✅ | Lowercase ('y') ✅
* **Expected Output:**
```markdown
Enter Password: Python123
Strong Password

````python
pwd=str(input("Enter Your Password"))
def valid_password(password):
    digit=any(char.isdigit() for char in password)
    upper=any(char.isupper() for char in password)
    lower=any(char.islower() for char in password)
    if digit and upper and lower and len(password)>=8:
        return "Strong Password"
    else:
        return "Weak Password"
print(valid_password(pwd))
````

## 📝 Challenge 08: Student Grade Report ⭐⭐⭐⭐

**Task:** Design a modular student grade management program that processes marks for five students using **function decomposition**—breaking down the problem so that each function handles a single, well-defined responsibility.

---

### 📋 Requirements

1. **Function Decomposition (4 Functions Required):**
   * `input_marks()`: Prompts the user to enter marks for 5 students and returns them as a list.
   * `calculate_average(marks_list)`: Takes a list of marks, calculates the mean score, and returns the average.
   * `calculate_grade(score)`: Accepts an individual student's score and returns a letter grade string based on grading rules (e.g., A, B, C, D, F).
   * `display_report(marks_list, average)`: Receives the marks and average score, calls `calculate_grade()` for each student, and prints a formatted report table.

2. **Program Execution Flow:**
   * Main execution calls `input_marks()` to gather student scores.
   * Calculates the overall average using `calculate_average()`.
   * Outputs the formatted summary via `display_report()`.

---

### 📥 Example Walkthrough

#### User Input (5 Students):
```text
Enter mark for Student 1: 85

Enter mark for Student 3: 68

Enter mark for Student 5: 55
========================================
         STUDENT GRADE REPORT           
========================================

Student 2 : Marks = 92 | Grade = A
Student 3 : Marks = 68 | Grade = C
Student 4 : Marks = 74 | Grade = B
Student 5 : Marks = 55 | Grade = D
----------------------------------------
Class Average : 74.80
========================================

````python
def students_list():
    students_marks=[]
    for i in range(0,5):
        student_number=i+1
        marks=float(input(f"Enter Marks of Student {student_number}"))
        students_marks.append(marks)
    return students_marks
student_marks=students_list()
def average(students_marks):
        avg=sum(students_marks)/len(students_marks)
        return avg
avg = average(student_marks)
def grades_calculation(students_marks):
    grades=[]
    for i in range(0,5):
        marks=students_marks[i]
        if marks>=80 and marks<=100:
            grade="A"
        elif marks>=70 and marks<=79:
            grade="B"
        elif marks>=60 and marks<=69:
            grade="C"
        elif marks>=50 and marks<=59:
            grade="D"
        else:
            grade="F"
        grades.append(grade)
    return grades
grade=grades_calculation(student_marks)
def report(grade,student_marks,avg):
    print(f"================================================")
    print(f"             STUDENT GRADE REPORT               ")
    print(f"================================================")
    for i in range(0,5):    
        print(f"Student {i+1}: Marks={student_marks[i]} | Grade={grade[i]}")
    print(f"------------------------------------------------")
    print(f"Class Average: {avg:.2f}")
    print(f"================================================")
report(grade,student_marks,avg)

 
````

## 📝 Challenge 11: Employee Salary Calculator ⭐⭐⭐⭐

**Task:** Build a modular salary processing system that collects an employee's details, computes dynamic bonus and tax deductions based on tiered salary thresholds, calculates total net pay, and displays an executive payroll report using **function decomposition**.

---

### 📋 Requirements

1. **Function Decomposition (5 Functions Required):**
   * `input_employee()`: Prompts the user to enter the employee's name and basic salary, returning both values.
   * `calculate_bonus(salary)`: Receives the basic salary and returns the calculated bonus amount based on bonus tier rules.
   * `calculate_tax(salary)`: Receives the basic salary and returns the calculated tax deduction based on tax tier rules.
   * `net_salary(salary, bonus, tax)`: Computes and returns the net salary using the formula:
     $$\text{Net Salary} = \text{Basic Salary} + \text{Bonus} - \text{Tax}$$
   * `display_salary_report(name, salary, bonus, tax, net)`: Accepts all calculated metrics and prints a structured employee payroll summary.

2. **Bonus Tiers:**
   * $\text{Salary} \ge 100,000 \longrightarrow 20\%$ of basic salary
   * $\text{Salary} \ge 70,000 \longrightarrow 10\%$ of basic salary
   * $\text{Otherwise} \longrightarrow 5\%$ of basic salary

3. **Tax Tiers:**
   * $\text{Salary} \ge 120,000 \longrightarrow 15\%$ of basic salary
   * $\text{Salary} \ge 80,000 \longrightarrow 10\%$ of basic salary
   * $\text{Otherwise} \longrightarrow 5\%$ of basic salary

---

### 📥 Example Walkthrough

#### User Input:
```text
Enter Employee Name: Sarah Khan
Enter Basic Salary: 105000


========================================
        EMPLOYEE SALARY REPORT          
========================================
Employee Name:      Sarah Khan
Basic Salary:       $105,000.00
Bonus Added:       +$21,000.00
Tax Deducted:      -$10,500.00
----------------------------------------
Net Salary:         $115,500.00
========================================

````python
def employee_input():
    name=str(input('Enter employee Name:'))
    salary=float(input("Enter emplyee Salary"))
    return name,salary
name, salary = employee_input()
def calculate_bonus(salary):
    if salary>=100000:
        bonus=salary*(20/100)
    elif salary>=80000:
        bonus=salary*(10/100)
    else:
        bonus=salary*(5/100)
    return bonus
bonus=calculate_bonus(salary)
def cal_tax(salary):
    if salary>=120000:
        tax=salary*(15/100)
    elif salary>=80000:
        tax=salary*(10/100)
    else:
        tax=salary*(5/100)
    return tax
tax=cal_tax(salary)
net_salary=salary+bonus-tax
def display_salary(name,salary,bonus,tax):
    print(f"=======================================")
    print(f"         EMPLOYEE SALARY REPORT        ")
    print(f"=======================================")
    print(f"{'Employee Name:':<20} {name.capitalize()}")
    print(f"{'Basic Salary:':<20} ${salary}")
    print(f"{'Bonus:':<20} +${bonus}")
    print(f"{'Tax:':<20} -${tax}")
    print(f"---------------------------------------")
    print(f"{'Net Salary:':<20} ${net_salary}")
    print(f"=======================================")
display_salary(name,salary,bonus,tax)
````
