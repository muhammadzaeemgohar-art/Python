## 📝 Challenge 02: Analyze Student Marks ⭐

**Task:** Write a Python program that takes a list of student marks entered by the user (or pre-defined) and computes key statistics using Python's built-in list functions: `max()`, `min()`, `sum()`, and `len()`.

---

### 📋 Requirements

1. **List Operations:** Use Python's built-in functions:
   * `max(marks)` to find the highest score.
   * `min(marks)` to find the lowest score.
   * `sum(marks)` to calculate the total sum of all scores.
   * `len(marks)` to count the total number of students.
2. **Calculations:** Compute the average score using the formula:
   $$\text{Average} = \frac{\text{sum(marks)}}{\text{len(marks)}}$$
3. **Output Formatting:** Print each calculated statistical metric cleanly formatted as shown in the example.

---

### 📥 Example Walkthrough

#### Given Marks List: `[78, 82, 91, 60, 95]`
* **Highest:** `95`
* **Lowest:** `60`
* **Total:** $78 + 82 + 91 + 60 + 95 = 393$
* **Average:** $\frac{393}{5} = 78.6$

#### Expected Output
```text
Highest Marks : 95
Lowest Marks  : 60
Average Marks : 78.6
Total Marks   : 393



````python
def marks_input():
    marks=[]
    for i in range(0,5):
        score=int(input(f"Enter Marks of Subject {i+1}:"))
        marks.append(score)
    return marks
def calculation(marks):
    highest_marks=max(marks)
    lowest_marks=min(marks)
    avg=sum(marks)/len(marks)
    total_marks=sum(marks)
    print(f"=======================================")
    print(f"           STUDENT REPORT              ")
    print(f"=======================================")
    for i in range(len(marks)):
        print(f"Subject {i+1} : {marks[i]}")
    print(f"---------------------------------------")
    print(f"Highest Marks : {highest_marks}")
    print(f"lowest Marks : {lowest_marks}")
    print(f"Average Marks : {avg:.2f}")
    print(f"Total Marks : {total_marks}")
    print(f"=======================================")
marks=marks_input()
calculation(marks)
````

## 📝 Challenge 05: Unique Subjects (Sets) ⭐⭐

**Task:** Prompt the user to enter 10 subjects, store them in a `set`, and display all unique subjects entered along with the total count of unique subjects to demonstrate automatic duplicate removal.

---

### 📋 Requirements

1. **Set Usage:** Create a `set` (or start with an empty set `subjects = set()`) to store the user inputs.
2. **Loop Input:** Use a `for` loop that runs **10 times** to ask the user to enter a subject name.
3. **Automatic Deduplication:** Add each entered subject to the set using `.add()`.
4. **Output Formatting:**
   * Iterate through the set and print each unique subject on a new line.
   * Display the total count of unique subjects using `len()`.

---

### 📥 Example Walkthrough

#### User Inputs (10 Total, with duplicates):
`Python`, `Math`, `Physics`, `Python`, `Statistics`, `Math`, `Python`, `Physics`, `Math`, `Statistics`

#### Deduplication Behavior:
* **Entered:** 10 items total
* **Unique Elements Extracted:** `{"Python", "Math", "Physics", "Statistics"}`

#### Expected Output
```text
Unique Subjects
---------------
Python
Math
Physics
Statistics

Total Unique Subjects: 4

````python
def subject_inputs():
    subjects=set()
    for subj in range(0,10):
        subject=str(input(f"Enter Subject {subj+1} name:")).title()
        subjects.add(subject)
    return subjects
subjects=subject_inputs()
print(f"Unique Subjects")
print(f"----------------------")
for i in subjects:
    print(i)
print(f"Total Unique Subjects:{len(subjects)}")    
````

````python
sentence=str(input("Enter Something:")).lower()
sentence=sentence.split()
frequency={}
for words in sentence:
    if words in frequency:
        frequency[words]+=1
    else:
        frequency[words]=1
print(f"Word Frequency Count")
print(f"---------------------------")
for key,values in frequency.items():
    print(f'{key.title()} : {values}')
````
