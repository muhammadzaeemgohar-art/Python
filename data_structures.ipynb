## 📝 Challenge 01: Student Marks List ⭐

**Task:** Create a program that initializes an empty list named `marks`, collects scores for 5 students using a `for` loop, appends each score to the list, and prints the resulting list.

---

### 📋 Requirements

1. **List Initialization:** Create an empty list called `marks`.
2. **Loop Input:** Use a `for` loop that runs **5 times** to prompt the user for student marks.
3. **Data Storage:** Store/append each mark into the `marks` list.
4. **Output Formatting:** Print the heading followed by the full list representation.

---

### 📥 Example Walkthrough

#### Sample Inputs
```text
Enter mark for student 1: 78
Enter mark for student 2: 82
Enter mark for student 3: 91
Enter mark for student 4: 65
Enter mark for student 5: 88

Student Marks
-------------
[78, 82, 91, 65, 88]

````python
marks=[]
for i in range(0,5):
    score=int(input(f"Enter Marks of Student {i+1}:"))
    marks.append(score)
print(f"Student Marks")
print(f"-----------------")
print(marks)
````

## 📝 Challenge Project: Library Management System ⭐⭐⭐⭐⭐

**Task:** Build a comprehensive, menu-driven Library Management System in Python. Each book is represented as a dictionary and all books are stored within a master list. The program relies on clean function decomposition, robust input validation, and dynamic updates to track book availability.

---

### 📋 Requirements

1. **Data Structures:**
   * **Book Record (Dictionary):** Contains the keys `"ID"` (int), `"Title"` (str), `"Author"` (str), and `"Available"` (bool).
   * **Library Database (List):** Holds all book dictionaries (e.g., `library = []`).

2. **Function Decomposition & Core Features:**
   * `add_book()`: Prompts for book details, ensures the ID is unique and positive, sets `"Available": True`, and appends the book to the library list.
   * `view_books()`: Displays all books in a formatted table layout showing ID, Title, Author, and Status (`Available` or `Borrowed`).
   * `search_book()`: Allows searching by Book ID or Title/Author substring.
   * `borrow_book()`: Updates a book's `"Available"` status to `False` if it is currently available. Displays an appropriate error if already borrowed or if the ID doesn't exist.
   * `return_book()`: Updates a book's `"Available"` status back to `True` if it is currently borrowed.
   * `delete_book()`: Removes a book record from the library by its ID after confirmation.
   * `main()` / Execution Loop: Drives the continuous interactive menu until option `7` is selected.

3. **Input Validation:**
   * Prevent crashes when users enter non-integer inputs for menu options or IDs (`ValueError` handling).
   * Prevent duplicate Book IDs when adding new items.
   * Guard against edge cases (e.g., trying to borrow an already borrowed book or returning a book that isn't checked out).
