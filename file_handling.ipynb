1. **Loop Input:** Use a `for` loop running **5 times** to ask the user for subject names.
2. **File Handling:** Open `subjects.txt` in write mode (`"w"`) using a `with open(...)` context manager block.
3. **Line Break Formatting:** Append a newline character (`\n`) to each subject string so each item is placed on a separate line.

---

### 📥 Example Walkthrough

#### User Inputs:
```text
Enter subject 1: Python
Enter subject 2: Statistics
Enter subject 3: Linear Algebra
Enter subject 4: Data Science
Enter subject 5: Programming

File Content Generated (subjects.txt):
Python
Statistics
Linear Algebra
Data Science
Programming



````python
subject=[]
for i in range(0,5):
    sub=input(f"Enter name of Subject {i+1}:")
    subject.append(sub)
with open("subject.txt","w") as file:
    file.write(f"Subjects List:\n")
    for i in range(len(subject)):
        file.write(f"{i+1}-{subject[i]}:\n")
````

### 📋 Requirements

1. **Data Model:** Each book record must track:
   * `ID` (Unique identifier, e.g., `101`)
   * `Title` (Book title)
   * `Author` (Author name)
   * `Available` (Boolean status: `True` or `False`)

2. **Menu Navigation:** Drive the program using a `while True` main loop supporting options `1` through `7`.

3. **Core Features:**
   * **1- Add Book:** Collect book details, set initial availability to `True`, and append to `books.txt`.
   * **2- View Books:** Read `books.txt` and display all books in a structured report. Handle cases where the file is empty or missing.
   * **3- Search Book:** Allow searching by **ID** or **Title** (case-insensitive substring/match).
   * **4- Borrow Book:** Locate the book by ID. If `Available` is `True`, update it to `False` and save changes to disk. Display an error if already borrowed.
   * **5- Return Book:** Locate the book by ID. If `Available` is `False`, update it to `True` and save changes to disk. Display an error if already returned/available.
   * **6- Delete Book:** Prompt for ID, remove the matching record, and rewrite the file with remaining records.
   * **7- Exit:** Exit the application cleanly.
