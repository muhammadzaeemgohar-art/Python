## 📝 Challenge 08: Student Marks Storage 🟢

**Task:** Write a Python program that prompts the user for student information (Name, Roll Number, and Marks in 3 subjects), then formats and appends this record to a file named `students.txt`. Subsequent entries should be appended so previous records are preserved.

---

1. **User Input:** Prompt for:
   * Student Name
   * Roll Number
   * Marks in Subject 1 (e.g., Python)
   * Marks in Subject 2 (e.g., Statistics)
   * Marks in Subject 3 (e.g., Mathematics)
2. **File Handling:**
   * Open `students.txt` in append mode (`"a"`) using `with open("students.txt", "a") as file:`.

3. **Data Preservation:** Ensure each execution or new record appends without overwriting existing data.
