# 📝 Question 09: Password Validation with `raise` 🟢

**Task:** Write a Python function `validate_password(password)` that checks password strength rules and uses `raise ValueError` to trigger custom exception messages when criteria are not met. Call the function inside a `try-except` block to test inputs.

---

### 📋 Requirements

1. **Function Definition:** Create `validate_password(password)`.
2. **Validation Rules (raise `ValueError`):**
   * **Length:** Raise an exception if `len(password) < 8`.
   * **Digit Check:** Raise an exception if the password contains no numbers (e.g., using `any(char.isdigit() for char in password)`).
   * **Uppercase Check:** Raise an exception if the password contains no uppercase letters (e.g., using `any(char.isupper() for char in password)`).
3. **Success State:** If all conditions pass, print `"Password is valid."`.
4. **Exception Handling:** Wrap the call to `validate_password()` inside a `try-except ValueError as e` block to display the error message (`e`) when validation fails.

---

### 📥 Example Walkthrough

#### Case 1: Short Password
```text
Enter password: Pass1
Error: Password must be at least 8 characters long.
```

#### Case 2: Missing Uppercase Letter
```text
Enter password: mypassword1
Error: Password must contain at least one uppercase letter.
```

#### Case 3: Valid Password
```text
Enter password: SecurePassword123
Password is valid.
```

---

### 💡 Enforcing Custom Logic Tip

<div style="padding: 15px; border-left: 5px solid #4CAF50; background-color: #e8f5e9; color: #1b5e20; border-radius: 4px;">
    <strong>Raising Custom Exceptions:</strong> Python won't automatically throw an error for domain-specific rules like password strength. Using <code>raise ValueError("Custom message")</code> allows you to interrupt program execution on invalid data and handle it cleanly in your <code>except</code> handler!
</div>

---

````python
try:
    password=input("Enter password:")
    if len(password)<8:
        raise ValueError("Password must be at least 8 characters long.")
    if not any(char.isupper() for char in password):
        raise ValueError("Password must contain at least one uppercase letter.")
    if not any(char.isdigit() for char in password):
        raise ValueError("Password must contain at least on digit.")
    print("Password is Valid.")
except ValueError as e:
    print(f"Error: {e}")
````

2. **Validation & Exception Handling (`raise ValueError`):**
   * Wrap input processing inside a `try` block.
   * Convert inputs to numbers (`float` or `int`).
   * Explicitly `raise ValueError("Marks must be between 0 and 100.")` if any score is less than 0 or greater than 100.
3. **Error Feedback:** Catch `ValueError` in an `except` block to display clear error messages for both non-numeric inputs and out-of-range marks.
4. **Success Output:** Print the recorded marks if all three inputs pass validation.

---

### 📥 Example Walkthrough

#### Case 1: Out-of-Range Mark
```text
Enter Python marks: 105
Error: Marks must be between 0 and 100.
```
#### Case 2: Non-Numeric Input
```text
Enter Python marks: 85

Error: Invalid input. Please enter a valid number.
```
#### Case 3: Valid Entry
```text
Enter Python marks: 88
Enter Mathematics marks: 92
Enter Statistics marks: 79

Marks Recorded Successfully:
- Python: 88.0
- Mathematics: 92.0
- Statistics: 79.0
```

````python
marks=[]
subjects=['Python','Mathematics','Statistics']
for i in range(0,3):
    while True:
        try:
            mark=int(input(f"Enter {subjects[i]} marks:"))
            if mark<0 or mark>100:
                raise ValueError("Marks must be between 0 to 100.")
            marks.append(mark)
            break
        except ValueError as e:
            print(f"Error: {e}")
print(f"Marks Recorded Successfully:")
for i in range(0,3):
    print(f" - {subjects[i]}: {marks[i]:.1f}")
````

### 💡 Multi-Exception File Handling Tip

> **Differentiating I/O Errors:** Catching distinct exceptions gives users precise feedback. `FileNotFoundError` occurs when a specified filename doesn't exist, whereas `PermissionError` occurs if your script attempts to edit read-only files or restricted system directories.

---

````python
import os
def read_file(file_name):
    with open(f"/home/zaeem/Python-test/Txt Files/{file_name}","r") as file:
        print("--------File Content---------")
        print(file.read())
        print("-----------------------------")
def write_file(file_name):
    with open(f"/home/zaeem/Python-test/Txt Files/{file_name}","w") as file:
        content=input("Enter File Content:")
        file.write(content)
        print(f"File Updated............")
def append_file(file_name):
    with open(f"/home/zaeem/Python-test/Txt Files/{file_name}","a") as file:
        content=input("Enter File Content:")
        file.write(content)
        print(f"File Updated............")
def validator(choice, file_name):
    if choice not in range(1, 5):
        raise ValueError("Invalid Input. Please Enter a valid menu number.")
    if choice == 1:
        path = f"/home/zaeem/Python-test/Txt Files/{file_name}"
        if not os.path.isfile(path):
            raise FileNotFoundError("The file was not found.")
while True:
        try:
            print(f"{'='*40}")
            print(f"{' '*10}SAFE FILE MANAGER")
            print(f"{'='*40}")
            print(f"1- Read File")
            print(f"2- Write File")
            print(f"3- Append File")
            print(f"4- Exit")
            print(f"{'='*40}")
            choice=int(input("Enter choice(1-4):"))
            if choice==4:
                print(f"Exiting File Manager................")
                break
            file_name=input("Enter filename:")
            validator(choice,file_name)
            if choice==1:
                read_file(file_name)
            elif choice==2:
                write_file(file_name)
            elif choice==3:
                append_file(file_name)
        except ValueError as e:
            print(f"Error: {e}")
        except FileNotFoundError as e:
            print(f"Error: {e}")
        except PermissionError as e:
            print(f"Error: {e}")
````

   * **2- View Books:** Catch `FileNotFoundError` if `books.txt` is missing. Catch `ZeroDivisionError` when calculating utilization statistics on an empty catalog.
   * **3- Search Book:** Search by ID or title. Handle missing files and blank search terms gracefully.
   * **4- Borrow Book:** Validate ID. Raise an error if the book is already borrowed (`Available: False`). Update disk state on success.
   * **5- Return Book:** Validate ID. Raise an error if the book is already available (`Available: True`). Update disk state on success.
   * **6- Delete Book:** Remove record by ID with validation for non-existent IDs and missing files.
   * **7- Exit:** Terminate application cleanly.
